## 6. Revision de inconsistencias y categorias

Se realizara una revision detallada de las variables categoricas para identificar diferencias de escritura, variaciones en el uso de mayusculas y posibles denominaciones equivalentes. Tambien se examinaran los nombres repetidos de las unidades deportivas para determinar si corresponden a registros diferentes.

```python
# Revisar todas las categorias de TIPO

print("Categorías originales de TIPO:")

display(df["TIPO"].value_counts().sort_index())
```

```python
# Identificar variaciones de escritura en BARRIO

barrios = df["BARRIO"].dropna()

grupos_barrios = (
    barrios.groupby(barrios.str.strip().str.casefold())
    .unique()
)

for grupo in grupos_barrios:
    if len(grupo) > 1:
        print(grupo)
```

```python
# Identificamos nombres de UNIDADES DEPORTIVAS repetidos

repetidos = df[
    df["UNIDADES DEPORTIVAS"].duplicated(keep=False)
].sort_values("UNIDADES DEPORTIVAS")

display(repetidos)
```

```python
# Identificar el valor maximo y los posibles valores atipicos de M2

print("Cinco areas mas grandes:")
display(
    df.nlargest(5, "M2")[
        ["item", "UNIDADES DEPORTIVAS", "M2", "BARRIO"]
    ]
)
```


## 7. Limpieza y normalizacion de los datos

A partir de los resultados obtenidos durante la exploracion inicial, se aplicaran operaciones de limpieza orientadas a mejorar la consistencia del conjunto de datos. Se normalizaran los formatos de las variables categoricas, se revisaran los valores faltantes y se conservaran los registros que no presenten evidencia suficiente de errores o duplicidad.

```python
# Crear una copia para realizar la limpieza

df_preparado = df.copy()

print("Copia de trabajo creada correctamente.")
```

```python
# Limpiar espacios al inicio y al final

columnas_texto = [
    "UNIDADES DEPORTIVAS",
    "TIPO",
    "CLASIFICACION",
    "BARRIO"
]

for columna in columnas_texto:
    df_preparado[columna] = (
        df_preparado[columna].str.strip()
    )

# Normalizar categorias

columnas_categoricas = [
    "TIPO",
    "CLASIFICACION",
    "BARRIO"
]

for columna in columnas_categoricas:
    df_preparado[columna] = (
        df_preparado[columna].str.upper()
    )

print("Normalizacion completada.")
```

```python
# Comprobamos las categorias despues de la normalizacion

print("Categorias de TIPO despues de la limpieza:")
display(df_preparado["TIPO"].value_counts())

print("\nCantidad de categorías de TIPO:")
print(df_preparado["TIPO"].nunique())

print("\nCantidad de barrios:")
print(df_preparado["BARRIO"].nunique())
```

```python
# Revisamos los valores faltantes

print("Valores faltantes despues de la limpieza:")
display(df_preparado.isnull().sum())
```

```python
# Convertir LOCALIDAD a tipo categorico

df_preparado["LOCALIDAD"] = (
    df_preparado["LOCALIDAD"].astype("category")
)

print(df_preparado.dtypes)
```

Evaluacion los valores atipicos.

```python
# Calcular cuartiles

q1 = df_preparado["M2"].quantile(0.25)
q3 = df_preparado["M2"].quantile(0.75)

# Calcular rango intercuartilico

iqr = q3 - q1

# Definir limites

limite_inferior = q1 - 1.5 * iqr
limite_superior = q3 + 1.5 * iqr

# Identificar posibles valores atipicos

atipicos = df_preparado[
    (df_preparado["M2"] < limite_inferior) |
    (df_preparado["M2"] > limite_superior)
]

print("Limite inferior:", limite_inferior)
print("Limite superior:", limite_superior)

print("Cantidad de posibles valores atipicos:", len(atipicos))

display(
    atipicos[
        ["item", "UNIDADES DEPORTIVAS", "M2", "BARRIO"]
    ].sort_values("M2", ascending=False)
)
```

## 8. Ingenieria de caracteristicas

Se crea una nueva variable denominada AREA_HA, utilizando los valores de la columna M2, que representa la superficie de cada escenario deportivo en metros cuadrados. Esta transformacion permite expresar las dimensiones de las instalaciones en hectareas, una unidad de medida que facilita la interpretación de superficies extensas.

```python
# Crear una variable con el area expresada en hectareas
df_preparado["AREA_HA"] = (df_preparado["M2"] / 10000).round(4)

# Mostrar algunos registros para comprobar la transformacion
display(
    df_preparado[["UNIDADES DEPORTIVAS", "M2", "AREA_HA"]].head()
)
```

## 9. Comparacion y aplicacion de los datos preparacion

A continuacion vamos a comprobar las diferencias entre el archivo original (df) y el preparado (df_preparado).

```python
# Comparacion de dimensiones
print("Datos originales:", df.shape)
print("Datos preparados:", df_preparado.shape)

# Comparacion de categorías
print("\nCategorias originales y preparadas:")
print("TIPO:", df["TIPO"].nunique(), "->", df_preparado["TIPO"].nunique())
print("BARRIO:", df["BARRIO"].nunique(), "->", df_preparado["BARRIO"].nunique())

# Comparacion de valores faltantes
print("\nValores faltantes originales:")
display(df.isnull().sum())

print("\nValores faltantes preparados:")
display(df_preparado.isnull().sum())
```
